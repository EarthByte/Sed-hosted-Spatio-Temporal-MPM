```python
import geopandas as gpd
from shapely import make_valid

# ~186 km in degrees at the equator (111.2 km/deg) → about 1.67°
# You can change this to 2.0 if you really want a full 2° buffer.
buffer_deg = 1.7      

recon_dir = os.path.join(input_directory, 'craton_lines')
buffer_dir = os.path.join(input_directory, 'P_craton_buffers')
os.makedirs(buffer_dir, exist_ok=True)


def clean_geom(g):
    if g is None:
        return g
    if g.is_empty:
        return g
    if not g.is_valid:
        g = make_valid(g)    # or g = g.buffer(0) as fallback
    return g


for t in range(0, 1801, 1):  # 1800, 1799, ..., 1
    

    shp_in = os.path.join(recon_dir, f"merged_segmented_craton_{t:04d}Ma.shp")
    if not os.path.exists(shp_in):
        
        continue

    gdf = gpd.read_file(shp_in)

    # Make sure we're in lat–lon (polar/geo) coordinates
    if gdf.crs is None:
        gdf = gdf.set_crs(epsg=4326)
    elif gdf.crs.to_epsg() != 4326:
        gdf = gdf.to_crs(epsg=4326)

    # 1) Buffer in degrees (full side + round ends)
    # cap_style=1 → round ends; join_style=1 → round corners
    gdf["buffer_geom"] = gdf.geometry.buffer(buffer_deg, cap_style=1, join_style=1)

    # 2) Clean invalid geometries
    gdf["buffer_geom"] = gdf["buffer_geom"].apply(clean_geom)

    # 3) Drop empty geometries
    gdf = gdf[~gdf["buffer_geom"].is_empty].copy()
    if gdf.empty:
        
        continue

    # 4) Merge all buffered pieces into a single feature
    merged_buffer = gdf["buffer_geom"].union_all()

    # 5) Save out as a one-feature shapefile
    out_gdf = gpd.GeoDataFrame(
        {"time": [t]}, geometry=[merged_buffer], crs="EPSG:4326"
    )

    shp_out = os.path.join(buffer_dir, f"craton_buffer_{t:04d}Ma.shp")
    out_gdf.to_file(shp_out)
```

```python
import os
import pygplates

# -----------------------------------------------------
# 1) RECONSTRUCT TO GEOJSON INSTEAD OF SHP
# -----------------------------------------------------

# Input shapefile (present-day / reference craton polygons)
craton_input_file = os.path.join('merged_segmented_craton.shp')

# Make an output folder to keep things tidy
out_dir = os.path.join(input_directory, 'P_craton_buffers_geojson')
os.makedirs(out_dir, exist_ok=True)

# Loop over times from 1800 Ma down to 1 Ma
for t in range(0, 1801, 1):   # 1800, 1799, ..., 1

    # Output GeoJSON file for this time slice
    out_geo = os.path.join(out_dir, f'merged_segmented_craton_{t:04d}Ma.geojson')
    
    # Reconstruct craton polygons and write to GeoJSON
    pygplates.reconstruct(
        craton_input_file,   # input craton shapefile
        rotation_model,      # your rotation model
        out_geo,             # output GeoJSON
        t                    # time (Ma)
    )

# -----------------------------------------------------
# 2) BUFFER + MERGE + EXPORT AS GEOJSON
# -----------------------------------------------------
import geopandas as gpd
from shapely import make_valid

# Buffer amount (in degrees)
buffer_deg = 1.7      

recon_dir = out_dir   # reconstructed .geojson folder
buffer_dir = os.path.join(input_directory, 'craton_buffers_merged_geojson')
os.makedirs(buffer_dir, exist_ok=True)

def clean_geom(g):
    if g is None:
        return g
    if g.is_empty:
        return g
    if not g.is_valid:
        g = make_valid(g)
    return g

for t in range(1800, 0, -1):  # 1800 → 1 Ma
    

    geo_in = os.path.join(recon_dir, f"merged_segmented_craton_{t:04d}Ma.geojson")
    if not os.path.exists(geo_in):
        
        continue

    gdf = gpd.read_file(geo_in)

    # Ensure CRS is EPSG:4326
    if gdf.crs is None:
        gdf = gdf.set_crs(epsg=4326)
    elif gdf.crs.to_epsg() != 4326:
        gdf = gdf.to_crs(epsg=4326)

    # 1) Buffer in degrees (round ends)
    gdf["buffer_geom"] = gdf.geometry.buffer(buffer_deg, cap_style=1, join_style=1)

    # 2) Clean invalid geometries
    gdf["buffer_geom"] = gdf["buffer_geom"].apply(clean_geom)

    # 3) Remove empties
    gdf = gdf[~gdf["buffer_geom"].is_empty].copy()
    if gdf.empty:
        
        continue

    # 4) Merge into one polygon
    merged_buffer = gdf["buffer_geom"].union_all()

    # 5) Export as GeoJSON
    out_gdf = gpd.GeoDataFrame({"time": [t]}, geometry=[merged_buffer], crs="EPSG:4326")

    geo_out = os.path.join(buffer_dir, f"craton_buffer_{t:04d}Ma.geojson")
    out_gdf.to_file(geo_out, driver="GeoJSON")
```
